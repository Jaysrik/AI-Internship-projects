```python
import pandas as pd
import matplotlib.pyplot as plt


file = r"C:\Users\Admin\Documents\prince_clg_db.xlsx"


columns = [
    "Student_ID", "Student_Name", "Gender", "Age", "Department",
    "Year", "Semester", "Section", "Email", "Phone",
    "City", "Marks", "CGPA", "Appears", "Admission_Year",
    "Course", "Hostel", "Scholarship", "Fees_Status",
    "Placement_Status"
]

students = pd.read_excel(
    file,
    sheet_name="Students",
    header=None
)


students.columns = columns


print("Data Shape:", students.shape)
print(students.head())


total_students = len(students)


total_departments = students["Department"].nunique()


average_mark = students["Marks"].mean()


average_attendance = students["Appears"].mean()


department_count = students["Department"].value_counts()


department_average = students.groupby(
    "Department"
)["Marks"].mean()

fees_values = students["Fees_Status"].value_counts()


hostel_count = students["Hostel"].value_counts()


fig = plt.figure(figsize=(14, 10))

# KPI 1     
plt.figtext(
    0.15, 0.90,
    f"Total Students\n{total_students}",
    ha="center",
    fontsize=14
)


#  KPI 2 
plt.figtext(
    0.38, 0.90,
    f"Departments\n{total_departments}",
    ha="center",
    fontsize=14
)

#  KPI 3
plt.figtext(
    0.62, 0.90,
    f"Average Mark\n{average_mark:.2f}",
    ha="center",
    fontsize=14
)

# KPI 4 
plt.figtext(
    0.85, 0.90,
    f"Avg Attendance\n{average_attendance:.2f}%",
    ha="center",
    fontsize=14
)


#  Department Chart
ax1 = fig.add_axes([0.08, 0.52, 0.40, 0.30])

department_count.plot(
    kind="bar",
    ax=ax1
)
ax1.set_title("Department-wise Students")
ax1.set_xlabel("Department")
ax1.set_ylabel("Students")

ax1.tick_params(
    axis="x",
    rotation=45
)


#  Average Marks Chart
ax2 = fig.add_axes([0.55, 0.52, 0.40, 0.30])

department_average.plot(
    kind="bar",
    ax=ax2
)

ax2.set_title("Department-wise Average Mark")
ax2.set_xlabel("Department")
ax2.set_ylabel("Average Mark")

ax2.tick_params(
    axis="x",
    rotation=45
)


# Fees Pie Chart 
ax3 = fig.add_axes([0.20, 0.08, 0.30, 0.30])

ax3.pie(
    fees_values.values,
    labels=fees_values.index,
    autopct="%1.1f%%",
    startangle=90
)

ax3.set_title("Fees Status")


#  Hostel Chart 
ax4 = fig.add_axes([0.58, 0.08, 0.30, 0.30])

hostel_count.plot(
    kind="bar",
    ax=ax4
)

ax4.set_title("Hostel-wise Students")
ax4.set_xlabel("Hostel")
ax4.set_ylabel("Students")

ax4.tick_params(
    axis="x",
    rotation=45
)


# Dashboard title
plt.suptitle(
    "College Student Database Dashboard",
    fontsize=20,
    fontweight="bold"
)

# Display dashboard
plt.show()
```
